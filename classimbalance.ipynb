To examine class imbalance in the Reached.on.Time_Yes/No column, means we are checking whether the dataset has a roughly equal number of "Yes" and "No" values—or if one class dominates.

1. Count the Class Distribution
Use value_counts() in pandas to see how many instances belong to each class:

```python
import pandas as pd

df = pd.read_csv('Train.csv')  
class_counts = df['Reached.on.Time_Y.N'].value_counts()
print(class_counts)
```

2. Visualize the Imbalance
A bar plot or pie chart helps make the imbalance visually clear:

```python
import matplotlib.pyplot as plt

# Bar plot
class_counts.plot(kind='bar', color=['skyblue', 'salmon'])
plt.title('Class Distribution: Reached on Time')
plt.xlabel('Reached on Time (Yes/No)')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()

# Optional: Pie chart
class_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['lightgreen', 'lightcoral'])
plt.title('Class Distribution: Reached on Time')
plt.ylabel('')
plt.show()
```

3. Calculate Class Proportions
This gives you a sense of how skewed the data is:

```python
class_proportions = df['Reached.on.Time_Y.N'].value_counts(normalize=True)
print(class_proportions)
```
